**Difference between sort & sorted**

```python
A = ["Oranges","Strawberry", "Mango"]
A.sort()
A
```

```python
A = ["Oranges","Strawberry", "Mango"]
B = A.sort()         # sort() can't be assigned
B
```

```python
A = ["Oranges","Strawberry", "Mango"]
B = sorted(A)
print(A)
B
```

**shallow copy**

```python
A = ["Oranges","Strawberry", "Mango"]
B = A #What happens: B = A does not create a new list. It makes B point to the exact same list object in memory as A.

#The result: Any modification made to A also affects B, because both variables are looking at the exact same data.
print(A)
print(B)
```

```python
print(A.pop())
print(A)
```

```python
B
```

```python
A = ["Oranges","Strawberry", "Mango"]
B = A[0:3] #What happens: Slicing A[0:3] creates a brand new list object in memory and copies the elements over into B.

#The result: A and B are completely independent lists. Modifying A will not affect B.
print(A)
print(B)
```

```python
print(A.pop())
print(A)
```

```python
B
```
