**Check whether a number is even or odd**

```python
x = 101

if x%2==0:
  print("even")
else:
  print("odd")
```

***Program which offers various discounts on purchase bills***

```python
shop_total = int(input("Enter the billed amount"))

if shop_total >= 5000:
  print("Wow! You won a voucher worth 1000")
elif shop_total >= 2500:
  print("Ohoo! You won a voucher worth 500")
else:
  print("Oops no vouchers")
```

**Iteration**

```python
#Iterate over a list of integers

list_1 = [10,20,30,40,50]
for num in list_1:
  print(num)
```

```python
#Iterate over a list of strings

str_1 = "Jharkhand"
for ch in str_1:
   print (ch)
```

```python
#Iterate over a dictionary
student_Data = {1: ["Sam",25],2: ["Sharma",21],3: ["Ravi", 23], 4: ["Shalu",23], 5: ["Sid", 22]}
#for i in student_Data:  <-- prints only keys
#for i in student_Data.items(): <-- prints both keys and values
#for i in student_Data.values():  #<-- prints only values
for i,j in student_Data.items():# <-- depending on i or j prints keys or values
   print(j)
```

```python
#Iterate over a range of values

for i in range(1,5):  #last one is not inclusive
   print(i)
```

**Comprehension**

```python
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia"]

list_2 = []

for i in list_1:
    list_2.append(i)

print(list_2)
```

```python
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia"]

list_2 = []

for i in list_1:
    list_2.append(len(i))

print(list_2)
```

```python

list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia"]

list_2 = [len(i) for i in list_1]
print(list_2)
```

```python
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia"]

d = {i: len(i) for i in list_1}

d
```

```python
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia"]

for i in list_1:
    d[i] = len(i)

d
```
