**Factorial**

```python

def fact(n):
  fact = 1

  for i in range (1,n+1):
      fact = fact*i
  return fact
```

```python
fact(5)
```


### Sum of all natural numbers till the number provided

```python
def sum(num):
  sum = 0
  for i in range(1,num+1):
    sum = sum + i
  return sum

```

```python
sum(10)
```
