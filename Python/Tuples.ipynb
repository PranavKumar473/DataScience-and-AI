**Indexing in tupling**

```python
t = ("Mumbai", 10, "Python", 100,"Delhi")
```

```python
print(t[0])     #gives element at index location 0
```

```python
t[2:]
```

```python
t[2:4]           #last index in exclusive
```

**Slicing**

Sorting a **tuple**

```python
t = (2,3,8,5,4,1)    #cannot sort a tuple (tuple->list->tuple)
new_list = sorted(t)
print(new_list)
```

```python
print(type(new_list))
```

```python
new_tuple = tuple(new_list)
print(new_tuple)
```
