### Accessing/Changing specific elements, rows & columns etc.

```python
a = np.array([[1,2,3,4,5,6,7,8],[8,7,6,5,4,3,2,1]])
print(a)
```

```python
a.ndim
```

```python
print(a[0,3])
```

```python
#Get a specific row from a numpy array
a[1,:]
```

```python
#Get a specific column froma numpy array
a[:,0]
```

```python
a[1,2:]
```

```python
a[1,2:5]
```

```python
a[0,5] = 70
```

```python
print(a[0,:])    #mistake: missed comma
```
