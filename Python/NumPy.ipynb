**Question #2**
Given a positive number 'n' greater than 2, create a NumPy array of size (nxn) with all zeros and ones such that the ones make a shape of "+"

```python
n = int(input("Enter your number: "))
```

```python
a = np.zeros((n,n),dtype='int')
```

```python
print(a)
```

```python
a[n//2,:]=1 #Row
a[:,n//2]=1.   #Column
print(a)
```

```python
b = np.array([[1,2,3,4],[50,60,70,80],[10,20,30,40],])
print(b)
```

```python
np.max(b,axis=0) #axis=0 → columns survive
                 #axis=1 → rows survive
```

```python
np.max(b,axis=1)
```

```python
np.min(b,axis=0)
```

```python
np.min(b,axis=1)
```

```python
np.sum(b,axis=0) #Sum vertically → one result for each column.
```

```python
np.sum(b,axis=1) #Sum horizontally ->one result for each row
```
