# **Dictionary**

1. Data in dictionary is stored as a **key: value pair**.

2. Key should not be mutable and values can be of any type.

**Dict = {"name":"John", 'age':10}**

Keys: name and age
Values: John and 10

```python
#creating a dictionary

d1 = {"India":"INR", "USA": "USD", "Hong Kong": "HKD"}
#Accessing value using keys
d1["India"]
```

```python
# Replacing the value for a key in a dictionary

d1["India"] = "Rs"
d1
```

```python
d1["Japan"] = "YEN"
d1
```

```python
d1["Hong Kong"] = "HK$"
d1
```

```python
#deleting a key value pair

del d1["Japan"]
d1
```

```python
#sorting a dictionary
sorted(d1)
```

```python
d1["Germany"]   #does not exist, hence error
```

```python
#values()
d1.values()
```

```python
#keys()
d1.keys()
```

```python
#get()
d1.get('India')
```

```python
d1.get('USA')
```

```python
d1.get('Hong Kong')
```

```python
d1["USA"] = "$"
d1
```

```python
#update
d1.update({"Hong Kong":"H$"})
d1
```

```python
d1["India"] = "INR"
d1
```
